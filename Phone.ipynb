Sort the phones in Entry Level , Midrange and Flagship phones in a new table

```python
import pandas as pd

df = pd.read_csv('phonedata.csv')
display(df.head())
print(df.info())
```

Based on the `df.info()` output, it looks like the 'Price' column is suitable for categorization. Now, let's define the price ranges and apply the categories.

```python
bins = [0, 20000, 50000, float('inf')]
labels = ['Entry Level', 'Midrange', 'Flagship']
df['Category'] = pd.cut(df['Selling Price'], bins=bins, labels=labels, right=True)
display(df.head())
display(df['Category'].value_counts())
```
